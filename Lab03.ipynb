## Task 3 · Clean the data: the six types of problem
Fix the problems **in this order** (text first, then duplicates):
- **(a) Missing values:** count them per column. Drop ads with no `Brand` or no `Model`; fill missing `Registered City`, `Fuel` and `Transaction Type` with `"Unknown"`; leave `Year` and `KMs Driven` as NaN. Print how many rows were removed.
- **(b) Formatting:** remove extra spaces and fix the case of city names (`.str.strip().str.title()`); remove `"Rs"` and commas from `Price`.
- **(c) Data types:** convert `Price` to `int64` and `Year` to `Int64`.
- **(d) Inconsistent categories:** `"Sukkar"` → `"Sukkur"` and `"Corrolla Altis"` → `"Corolla Altis"`. Print the number of different cities before and after.
- **(e) Duplicates:** count exact duplicate ads, drop them, print the new row count.
- **(f) Invalid values:** set `KMs Driven` to NaN when it is below 100, above 1,000,000, or the placeholder 12345 / 123456. Drop ads with the fake prices 12,345,678 or 87,654,321, and Suzuki/Daihatsu ads above 5,000,000.
- **(g) Self-check:** your final shape must be **(20484, 9)**.

```python
# T3
# (a) missing values
print("T3a: missing values per column")
print(df.isna().sum())
rows_before = len(df)
df = df.dropna(subset=["Brand", "Model"])
print("T3a: rows removed =", rows_before - len(df))
for col in ["Registered City", "Fuel", "Transaction Type"]:
    df[col] = df[col].fillna("Unknown")

# (b) formatting
df["Registered City"] = df["Registered City"].str.strip().str.title()
df["Price"] = (df["Price"].str.replace("Rs", "", regex=False)
                          .str.replace(",", "", regex=False)
                          .str.strip())

# (c) data types
df["Price"] = df["Price"].astype("int64")
df["Year"] = df["Year"].astype("Int64")
print("T3c:")
print(df[["Price", "Year"]].dtypes)

# (d) inconsistent categories
print("T3d: cities before =", df["Registered City"].nunique())
df["Registered City"] = df["Registered City"].replace({"Sukkar": "Sukkur"})
df["Model"] = df["Model"].replace({"Corrolla Altis": "Corolla Altis"})
print("T3d: cities after  =", df["Registered City"].nunique())

# (e) duplicates
print("T3e: exact duplicates =", df.duplicated().sum())
df = df.drop_duplicates()
print("T3e: rows after dropping =", len(df))

# (f) invalid values
bad_km = (df["KMs Driven"] < 100) | (df["KMs Driven"] > 1_000_000) | df["KMs Driven"].isin([12345, 123456])
print("T3f: KMs set to NaN =", bad_km.sum())
df.loc[bad_km, "KMs Driven"] = np.nan

fake_price = df["Price"].isin([12_345_678, 87_654_321])
too_dear = df["Brand"].isin(["Suzuki", "Daihatsu"]) & (df["Price"] > 5_000_000)
print("T3f: ads dropped for bad price =", (fake_price | too_dear).sum())
df = df[~(fake_price | too_dear)]

# (g) self-check
print("T3g: final shape =", df.shape)
assert df.shape == (20484, 9)
```

## Task 4 · Analyse
- **(a)** New column `Price_lakh` = price in lakh rupees (÷ 100,000).
- **(b)** New column `Age` = 2020 − `Year`.
- **(c)** The 5 most common brands.
- **(d)** The 5 cities with the most ads.
- **(e)** Median `Price_lakh` for each fuel type, highest first.
- **(f)** Median `Price_lakh` of a **Corolla GLI** for each year 2010–2018. How much value does it lose per year?
- **(g)** The 5 most expensive ads. Does one of them look wrong? Why?
- **(h)** Number of ads and median price for Cash vs Installment/Leasing. Interpret in 1–2 lines.

```python
# T4
# (a), (b)
df["Price_lakh"] = df["Price"] / 100_000
df["Age"] = 2020 - df["Year"]
print("T4b:")
print(df[["Price", "Price_lakh", "Year", "Age"]].head())

# (c)
print("T4c: top 5 brands")
print(df["Brand"].value_counts().head(5))

# (d)
print("T4d: top 5 cities")
print(df["Registered City"].value_counts().head(5))

# (e)
print("T4e: median Price_lakh by fuel")
print(df.groupby("Fuel")["Price_lakh"].median().sort_values(ascending=False))

# (f)
gli = df[(df["Model"] == "Corolla GLI") & df["Year"].between(2010, 2018)]
gli_median = gli.groupby("Year")["Price_lakh"].median()
print("T4f: Corolla GLI median Price_lakh by year")
print(gli_median)
print("T4f: average change per year (lakh) =", round(gli_median.diff().mean(), 2))
print("T4f: average change per year (%)    =", round(gli_median.pct_change().mean() * 100, 1))

# (g)
print("T4g: 5 most expensive ads")
print(df.nlargest(5, "Price")[["Brand", "Model", "Year", "KMs Driven", "Price", "Price_lakh"]])

# (h)
print("T4h:")
print(df[df["Transaction Type"].isin(["Cash", "Installment/Leasing"])]
        .groupby("Transaction Type")["Price_lakh"].agg(ads="count", median="median"))
```

**T4f answer:** A Corolla GLI gets cheaper as it gets older. Going from 2018 back to 2010 the median price falls from about 23.2 lakh to 12.25 lakh, which is roughly 1.4 lakh (about 8%) lost for each extra year of age.

**T4g answer:** Yes, the 2005 Land Cruiser at Rs 52,000,000 (520 lakh) looks wrong. A 20-year-old car with 97,000 km is listed above a 2015 Land Cruiser at 420 lakh, which does not fit the age/price pattern from T4f. It is probably a typing mistake (an extra zero or digit) or a joke price. The odometer values on the top ads (32222, 22222, 3222) also look typed rather than real.

**T4h answer:** Cash ads are the vast majority (19,634 vs 850) and have a median of 6.9 lakh, while Installment/Leasing ads have a median of only 3.0 lakh. So installment deals are rarer and are mostly attached to cheaper cars, which suggests that buyers of cheaper cars are the ones who need to pay in parts, or that expensive cars are mostly sold for cash.
