## Task 6
**What is the main difference between a tuple and a list?**

**Answer:** A list is *mutable*: its items can be changed, added or removed after creation (`my_list[0] = 5`, `append`, etc.). A tuple is *immutable*: once created it cannot be modified. Lists use square brackets `[]`, tuples use parentheses `()`. Tuples are therefore used for fixed data (e.g. an image's `(height, width)`) and are slightly faster and safer to use as dictionary keys.

```python
my_list = [1, 2, 3]
my_tuple = (1, 2, 3)

my_list[0] = 99          # works
print("Q6: list after change :", my_list)

try:
    my_tuple[0] = 99     # fails
except TypeError as e:
    print("Q6: tuple error       :", e)
```

## Task 8
Function: return `True` if the word "dog" is in a string (case-insensitive).

```python
def find_dog(text):
    """True if 'dog' appears in text, ignoring case."""
    return 'dog' in text.lower()

print("Q8:", find_dog("Is there a dog here?"))
print("Q8:", find_dog("Is there a DOG here?"))
print("Q8:", find_dog("No pets here"))
```

## Task 9
Function: count how many times "dog" occurs in a string (case-insensitive).

```python
def count_dog(text):
    """Count occurrences of 'dog' in text, ignoring case."""
    return text.lower().count('dog')

print("Q9:", count_dog("This dog runs faster than the other dog"))
print("Q9:", count_dog("Dog, DOG and dog"))
```

## Task 10
`caught_speeding(speed, is_birthday)`:
- speed ≤ 60 → "No Ticket"
- 61–80 → "Small Ticket"
- ≥ 81 → "Big Ticket"
- On your birthday, the limits are 5 higher.

```python
def caught_speeding(speed, is_birthday):
    """Return the ticket level for a given speed."""
    allowance = 5 if is_birthday else 0
    if speed <= 60 + allowance:
        return "No Ticket"
    elif speed <= 80 + allowance:
        return "Small Ticket"
    else:
        return "Big Ticket"

print("Q10:", caught_speeding(60, False))   # No Ticket
print("Q10:", caught_speeding(65, False))   # Small Ticket
print("Q10:", caught_speeding(65, True))    # No Ticket
print("Q10:", caught_speeding(85, True))    # Small Ticket
print("Q10:", caught_speeding(86, True))    # Big Ticket
print("Q10:", caught_speeding(81, False))   # Big Ticket
```

## Bonus Challenges

### Bonus 1: `count_dog` that also handles punctuation ("dog," and "dog!")

```python
import string

def count_dog_clean(text):
    # strip punctuation, then count whole word 'dog'
    cleaned = text.lower().translate(str.maketrans('', '', string.punctuation))
    return cleaned.split().count('dog')

print("Bonus1:", count_dog_clean("Dog! I saw a dog, and another dog."))
```

### Bonus 2: `is_palindrome(word)` using slicing

```python
def is_palindrome(word):
    word = word.lower()
    return word == word[::-1]

print("Bonus2:", is_palindrome("Level"), is_palindrome("Python"))
```
